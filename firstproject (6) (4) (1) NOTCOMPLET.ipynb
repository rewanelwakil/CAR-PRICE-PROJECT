```python
Q1 = dat['selling_price'].quantile(0.25)   # الربع الأول
Q3 = dat['selling_price'].quantile(0.75)   # الربع الثالث
IQR = Q3 - Q1                              # المدى بين الأرباع

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("الحد الأدنى:", lower_bound)
print("الحد الأعلى:", upper_bound)
```

```python
import datetime
current_year = datetime.datetime.now().year
dat['car_age'] = current_year - dat['year']
dat = dat.drop(columns=['year'])


print( dat.columns.tolist())
```

```python
#splliting
X = dat.drop(columns=['selling_price'])  
Y = dat['selling_price']                    

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
predicted = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print(f"R2 Score: {r2_score(Y_test, predicted):.2f}")
print(f"MAE: {mean_absolute_error(Y_test, predicted):.2f}")
```

```python
import matplotlib.pyplot as plt

actual_50 = Y_test.values[:50]
predicted_50 = predicted[:50]
sorted_order = actual_50.argsort()

plt.figure(facecolor='white', figsize=(10, 6))
plt.plot(range(1, 51), actual_50[sorted_order], 'o-', color='blue', label='Actual', linewidth=2, markersize=5)
plt.plot(range(1, 51), predicted_50[sorted_order], 's--', color='red', label='Predicted', linewidth=2, markersize=5)
plt.title("Car Price Prediction Using Linear Regression", fontsize=13, fontweight='bold')
plt.xlabel("Sample")
plt.ylabel("Selling Price")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
