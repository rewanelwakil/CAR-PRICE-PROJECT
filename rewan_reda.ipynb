```python
#splliting
X = dat.drop(columns=['selling_price'])  
Y = dat['selling_price']                    

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
dat = dat.drop(columns=['fuel','seller_type','transmission','owner'])
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error

model_rf = RandomForestRegressor(n_estimators=200, max_depth=15, min_samples_split=5, random_state=42)
model_rf.fit(X_train, Y_train)
predicted_rf = model_rf.predict(X_test)

print(f"R2 Score: {r2_score(Y_test, predicted_rf):.2f}")
print(f"MAE: {mean_absolute_error(Y_test, predicted_rf):.2f}")
```

```python
actual_50 = Y_test.values[:50]
predicted_50 = predicted_rf[:50]

sorted_order = actual_50.argsort()
actual_sorted = actual_50[sorted_order]
predicted_sorted = predicted_50[sorted_order]

plt.figure(facecolor='white', figsize=(10, 6))
plt.plot(range(1, len(actual_sorted)+1), actual_sorted, 'o-', color='blue', label='Actual', linewidth=2, markersize=5)
plt.plot(range(1, len(predicted_sorted)+1), predicted_sorted, 's--', color='red', label='Predicted', linewidth=2, markersize=5)
plt.title("Car Price Prediction Using Random Forest", fontsize=13, fontweight='bold')
plt.xlabel("Sample")
plt.ylabel("Selling Price")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
